```python
from box import ConfigBox
```

```python
d2=ConfigBox({'key':'value','key1':'value1'})
```

```python
d2.key  #now it will give value because ConfigBox is a class which has attribute key
#ConfigBox is a class which is used to convert dictionary into object so that we can access the values using dot(.) operator instead of using square brackets([])
```

```python
def get_prodcut(x:int,y:int)->int:
    return x*y
```

```python
get_prodcut(2,3)
```

```python
get_prodcut(2,'3')
```

```python
from ensure import ensure_annotations
```

```python
@ensure_annotations
def get_product(x:int,y:int)->int:
    return x*y
```

```python
get_product(x=2,y=3)
```

```python
get_product(x=2,y="3")  #now it will give error because we have used ensure_annotations decorator which will check the type of the arguments and return type of the function. If the type is not correct then it will give error.
```
